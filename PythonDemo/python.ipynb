## Python Notes 
### List
- Lists are very similar to arrays.
- It is one of the most used datatype in Python and is very flexible.
- List is mutable - list once created can be modified

### Dynamic list


```python
my_list = []
my_value = int(input("Enter number of values :"))

for i in range(my_value):
    value = input(f"Enter value at index {i+1}:")
    my_list.append(value)
    
print(my_list)
```

### Dictionary 
- Dictionarise are kind of hash table type.
- Dictionarise are enclosed by curly braces ({}).


```python
mydic = {
    "A": "Apple",
    50 : "Schine",
    "P": "Python",
    99.99: 99
        }

for i in mydic:
    print(f"Key is {i} Value is {mydic[i]}")
```
