```python
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Java Script</title>
    <script>
        document.write("<table border='1'>")
        for (var i = 0; i < 1000; i++){
            document.write("<tr>")
                if(i % 2 == 0){
                    document.write("<td bgcolor='red'>" + i + "</td>")
                }
                else{
                    document.write("<td bgcolor='yellow'>" + i + "</td>")
                }
                document.write("</tr>")
        }
        document.write("</table>")
    </script>
</head>
<body>
</body>
</html>
```

```python
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Array</title>
    <script>
        var myarr = ["C","C++","Python","Git","Javascript","Php","C#"]
        for (var i = 0; i<myarr.length; i++){
            document.write("<br/>")
            document.write(myarr[i])
        }

        for(i in myarr){
            document.write("<br/>Index is " + i + " Value is " + myarr[i])
        }

        for(i of myarr){
            document.write("<br/> Value is " + i)
        }

        myarr.forEach((value,index,arr)=>{
            document.write("<br/> Value is " + value)
        })
    </script>
</head>
<body>
    
</body>
</html>
```

```python
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Input</title>
    <script>
        function demo1() {
            alert("Welcome to JS")
        }

        function demo2() {
            var a = confirm("Are you sure?")
            if ( a == true){
                alert("Record Deleted")
            }
        }

        function demo3() {
            var a = prompt("Enter No1")
            var b = prompt("Enter No2")
            var c = parseInt(a) + parseInt(b)
            alert("Sum is " + c)
        }
    </script>
</head>
<body>
    <input type="button" onclick="demo1()" value="Button 1">
    <input type="button" onclick="demo2()" value="Button 2">
    <input type="button" onclick="demo3()" value="Button 3">
</body>
</html>
```
